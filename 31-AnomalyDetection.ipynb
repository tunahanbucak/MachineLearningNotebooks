```python
# Anomali tespitinin burada işe yarayıp yaramadığını görmek için modelleri karşılaştıralım.
```

```python
from sklearn.ensemble import IsolationForest
```

```python
#burda Isolation Forest modelini kullanıyoruz. Bu model, anomali tespiti için kullanılan bir algoritmadır.
# n_estimators parametresi, kaç tane izolasyon ağacı kullanılacağını belirler. 
# contamination parametresi, veri setindeki anomali oranını belirtir. 
# max_samples parametresi, her bir izolasyon ağacında kaç tane örnek kullanılacağını belirler.
#  max_features parametresi, her bir izolasyon ağacında kaç tane özellik kullanılacağını belirler.
#  random_state parametresi, rastgele sayı üreteci için bir başlangıç değeri belirler.
#  n_jobs parametresi, paralel işlem sayısını belirtir.
iso = IsolationForest(
    n_estimators=200, 
    contamination=fraud_ratio,
    max_samples="auto", 
    max_features=1.0, 
    random_state=42,
    n_jobs=-1
)
```

```python
#X değişkeni, Class sütunu hariç tüm sütunları içeren bir DataFrame'dir. 
# Bu sayede, modelleme aşamasında Class sütunu kullanılmaz ve model sadece diğer özellikleri kullanarak anomali tespiti yapar. 
# iso.fit ise, modelin X verisi üzerinde eğitilmesini sağlar.
#  Bu sayede, model anomali tespiti için gerekli olan bilgileri öğrenir ve daha sonra yeni veriler üzerinde anomali tespiti yapabilir.
X = df.drop("Class", axis=1)

iso.fit(X)
```

```python
#iso.predict(X) fonksiyonu, modelin X verisi üzerinde anomali tespiti yapmasını sağlar. 
# iso_raw sütunu, modelin tahmin ettiği anomali değerlerini içerir. 
# Bu sütun, 1 ve -1 değerlerini alır. 1 değeri, normal bir veri noktasını temsil ederken, -1 değeri anomali bir veri noktasını temsil eder. 
# Bu sayede, modelin tahmin ettiği anomali değerlerini df DataFrame'ine ekleyebiliriz.
df["iso_raw"] = iso.predict(X)
```

```python
df['iso_raw'].unique()
```

```python
#1  → normal
#-1 → anomaly
```

```python
#iso_pred sütunu, modelin tahmin ettiği anomali değerlerini 0 ve 1 değerlerine dönüştürür. 
# burda değişim yaptık çünkü 1 normal ve -1 anomaliyi temsil ediyor. biz ise 0 normal ve 1 anomaliyi temsil etmesini istiyoruz.
df["iso_pred"] = df["iso_raw"].map({1:0, -1:1})
```

```python
df.head()   
```

```python
df.tail()
```

```python
from sklearn.model_selection import train_test_split
```

```python
# X değişkeni, Class, iso_raw ve iso_pred sütunları hariç tüm sütunları içeren bir DataFrame'dir. y ise, Class sütununu içeren bir Series'dir. 
# X_train, X_test, y_train, y_test değişkenleri, veri setini eğitim ve test setlerine ayırmak için kullanılır. 
# test_size parametresi, test setinin boyutunu belirler. 
# stratify parametresi, veri setinin sınıf dağılımını korumak için kullanılır.
#  random_state parametresi, rastgele sayı üreteci için bir başlangıç değeri belirler.

X = df.drop(["Class","iso_raw","iso_pred"], axis=1)
y = df["Class"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    stratify=y,
    random_state=42
)
```

```python
#iso_test_pred değişkeni, test setindeki örnekler için modelin tahmin ettiği anomali değerlerini içerir. 
# X_test.index kullanılarak, test setindeki örneklerin indeksleri alınır ve df DataFrame'inde bu indekslere karşılık gelen iso_pred değerleri seçilir.
#  Bu sayede, test setindeki örnekler için modelin tahmin ettiği anomali değerlerini elde edebiliriz.
iso_test_pred = df.loc[X_test.index, "iso_pred"]
```

```python
#recall skoru, modelin anomali (fraud) olan örnekleri ne kadar iyi tespit ettiğini gösterir. 
# burda recall skoru, modelin anomali olan örnekleri doğru bir şekilde tespit etme yeteneğini ölçer. 
from sklearn.metrics import classification_report

print(classification_report(y_test, iso_test_pred))
```

```python
# Geri çağırma oranı %24 ile biraz düşük görünüyor, ancak verilerin yalnızca %0,17'sinin anormallik olabileceğini düşünürsek.
# Aslında iyi de olabilir. Şimdi asıl bakmamız gerekenlere bakalım.
```

```python
#decision_function, modelin her bir örnek için anomali skorunu hesaplamasını sağlar.
df["anomaly_score"] = iso.decision_function(X)
```

```python
#kdplot, veri dağılımını görselleştirmek için kullanılan bir yöntemdir. anomaly_score sütunu, modelin her bir örnek için hesapladığı anomali skorlarını içerir.
# fraud ve normal sınıfların anomali skorlarının dağılımını karşılaştırmak için kullanılır.
sns.kdeplot(
    df[df["Class"]==0]["anomaly_score"],
    label="Normal"
)

sns.kdeplot(
    df[df["Class"]==1]["anomaly_score"],
    label="Fraud"
)

plt.legend()
plt.title("Anomaly Score Distribution")
plt.show()
```

```python

# Anormallik puanı, anormalliklerde sola, normal durumlarda sağa doğru dağılır ve bu tam olarak şöyle olmalıdır.
```

```python

# En şüpheli %1'lik işlemlere bakalım
```

```python
df["anomaly_score"].quantile(0.01)
```

```python
#Eşik değerini 0.0017 yerine bu değere çıkarırsak ne olacağını görelim.
```

```python
#threshold, anomali skorlarının %1'lik alt sınırını temsil eder. Bu eşik değeri, anomali tespiti için kullanılacak olan sınır değerini belirler.
#  quantile(0.01) fonksiyonu, anomali skorlarının %1'lik alt sınırını hesaplar ve bu değeri threshold değişkenine atar.
threshold = df["anomaly_score"].quantile(0.01)

df["iso_custom"] = (df["anomaly_score"] < threshold).astype(int)
```

```python
print(classification_report(df["Class"], df["iso_custom"]))
```

```python
# Gördüğünüz gibi, geri çağırma oranı aslında artıyor ve hassasiyet %1 düşüyor. Gerçek hayatta istediğimiz şey de bu.
```

```python

#bu grafik, anomali skorlarının dağılımını ve karar eşik değerini görselleştirir.
sns.kdeplot(df[df["Class"]==0]["anomaly_score"], label="Normal")
sns.kdeplot(df[df["Class"]==1]["anomaly_score"], label="Fraud")

plt.axvline(threshold, color="red", linestyle="--",
            label="Decision Threshold")

plt.legend()
plt.title("Anomaly Score Separation")
plt.show()
```

```python
#sort_values() fonksiyonu, df DataFrame'ini anomaly_score sütununa göre sıralar.
df_sorted = df.sort_values("anomaly_score")
```

```python
# o zaman en iyi %1'e gireriz
k = int(len(df_sorted) * 0.01)

top_k = df_sorted.head(k)
```

```python
top_k
```

```python
top_k["Class"].value_counts()
```

```python
top_k["Class"].sum()
```

```python
#frauds_captured değişkeni, top_k DataFrame'inde fraud olan örneklerin sayısını temsil eder.
# total_frauds değişkeni, df DataFrame'inde fraud olan örneklerin
#recall_at_k değişkeni, top_k DataFrame'inde fraud olan örneklerin sayısının, df DataFrame'inde fraud olan örneklerin sayısına oranını temsil eder.

frauds_captured = top_k["Class"].sum()
total_frauds = df["Class"].sum()

recall_at_k = frauds_captured / total_frauds

print(recall_at_k)
```

```python
recall_at_k
```

```python
# Bu, işlemlerin yalnızca %1'ini değerlendirerek dolandırıcılığın neredeyse %60'ını ortaya çıkardığımız anlamına geliyor.
```

```python

# Bunu şu şekilde genişletebiliriz
```

```python
#percentages değişkeni, 0.001 ile 0.05 arasında 30 eşit aralıklı değer içeren bir NumPy dizisidir. 
# recalls listesi, her bir yüzde için geri çağırma oranlarını saklamak için kullanılır.
# total_frauds değişkeni, df DataFrame'inde fraud olan örneklerin sayısını temsil eder. 
# perdencateg ise, her bir yüzde için geri çağırma oranlarını hesaplamak için kullanılan bir döngüde kullanılacak olan değişkeni temsil eder.
percentages = np.linspace(0.001, 0.05, 30)
recalls = []

total_frauds = df["Class"].sum()

for p in percentages:
    k = int(len(df_sorted) * p)
    captured = df_sorted.head(k)["Class"].sum()
    recalls.append(captured / total_frauds)
```

```python
plt.plot(percentages*100, recalls)

plt.xlabel("Investigated Transactions (%)")
plt.ylabel("Fraud Recall")
plt.title("Top-K Capture Curve (Isolation Forest)")

plt.show()
```

```python
from xgboost import XGBClassifier
```

```python
#burda XGBClassifier modelini kullanıyoruz. Bu model, denetimli öğrenme algoritmalarından biridir ve sınıflandırma problemlerinde yaygın olarak kullanılır.
# scale_pos_weight parametresi, dengesiz veri setlerinde sınıf ağırlıklarını ayarlamak için kullanılır.
X = df.drop(["Class", "iso_raw", "iso_pred", "iso_custom"], axis=1)
# anomaly_score is already in X 

y = df["Class"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, stratify=y, random_state=42
)

xgb = XGBClassifier(
    scale_pos_weight=len(y_train[y_train==0]) / len(y_train[y_train==1]),
    random_state=42, n_jobs=-1
)
#scale_pos_weight to handle unbalanced dataset

xgb.fit(X_train, y_train)
print(classification_report(y_test, xgb.predict(X_test)))
```
