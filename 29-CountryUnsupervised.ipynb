```python
import math

def plot_all_histograms(df, title_prefix=""):
    num_cols = df.select_dtypes(include=[np.number]).columns
    n_cols = 3
    n_rows = math.ceil(len(num_cols) / n_cols)

    plt.figure(figsize=(5 * n_cols, 4 * n_rows))

    for i, col in enumerate(num_cols, 1):
        plt.subplot(n_rows, n_cols, i)
        sns.histplot(df[col], kde=True, bins=30)
        plt.title(f"{title_prefix}{col}")
        plt.xlabel("")
        plt.ylabel("")

    plt.tight_layout()
    plt.show()
```

```python
plot_all_histograms(df)
```

```python
from sklearn.decomposition import PCA
```

```python
# pca ile boyut indirgeme işlemi yapacağız. 
# PCA, verinin varyansını koruyarak boyutunu azaltır ve veriyi daha anlamlı bir şekilde temsil eder. 
# Bu sayede görselleştirme ve modelleme işlemleri daha etkili hale gelir.
pca = PCA()
```

```python
pca_df2 = pd.DataFrame(pca.fit_transform(df2))
```

```python
pca_df2
```

```python
#burda, pca.explained_variance_ratio_ ile her bir bileşenin toplam varyansa katkısını görebiliriz. 
# Bu sayede hangi bileşenlerin daha fazla bilgi taşıdığını anlayabiliriz.
pca.explained_variance_ratio_
```

```python
# burdaki grafikte, PCA'nın her bir bileşenin toplam varyansa katkısını gösteren bir adım grafiği ve bir çizgi grafiği çiziyoruz.
#  Bu sayede hangi bileşenlerin daha fazla bilgi taşıdığını görselleştirebiliriz.
plt.step(list(range(1,10)), np.cumsum(pca.explained_variance_ratio_))
plt.plot(np.cumsum(pca.explained_variance_ratio_))
plt.ylabel("Variance Covered")
plt.title("Variance Covered")
plt.show()
```

```python

pca_df2 = pca_df2.drop(columns = [3,4,5,6,7,8])
```

```python
pca_df2
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
# burada KMeans algoritmasını kullanarak farklı k değerleri için model oluşturuyoruz.
# Her bir k değeri için modeli eğitip, inertia değerini (WCSS) hesaplıyoruz ve bu değerleri wcss listesine ekliyoruz. 
# Bu sayede, farklı k değerleri için modelin performansını karşılaştırabiliriz ve en uygun k değerini belirleyebiliriz.
wcss = []
for k in range(1,11):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(pca_df2)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1,11),wcss)
plt.xticks(range(1,11))
plt.xlabel("number of clusters")
plt.ylabel("WCSS")
plt.show()
```

```python
# KMeans algoritmasını kullanarak 3 küme için model oluşturuyoruz. 
# Daha sonra bu modeli PCA ile boyut indirgeme işlemi yapılmış veriye fit ediyoruz. Bu sayede veriyi 3 farklı kümeye ayırıyoruz.
model = KMeans(n_clusters=3)
model.fit(pca_df2)
```

```python
model.labels_
```

```python
# labels değişkeni, KMeans modelinin her bir veri noktasını hangi kümeye atadığını gösterir. Bu sayede veriyi 3 farklı kümeye ayırabiliriz.
labels = model.labels_
```

```python
# burada silhouette_score fonksiyonunu kullanarak modelin performansını değerlendiriyoruz. 
# Silhouette skoru, kümeleme algoritmasının ne kadar iyi çalıştığını gösteren bir metriktir. 
# Skor 1'e yakınsa, kümeler iyi ayrılmış demektir; skor 0'a yakınsa, kümeler birbirine çok yakın demektir; 
# skor -1'e yakınsa, veriler yanlış kümelere atanmış demektir.
silhouette_score(pca_df2,labels)
```

```python
df['Class'] = labels
```

```python
df
```

```python
#burda child_mort ve income değişkenlerini Class değişkenine göre boxplot ile görselleştiriyoruz. 
# Daha detaylı açıklama: Boxplot, verinin dağılımını ve özet istatistiklerini görselleştiren bir grafiktir.
# biz de şunları görebiliriz: her bir Class için child_mort ve income değişkenlerinin dağılımını, medyanını, çeyreklerini ve olası aykırı değerleri.
fig, ax = plt.subplots(nrows = 1, ncols= 2 , figsize=(15,5))

plt.subplot(1,2,1)
sns.boxplot(data=df, x="Class", y="child_mort")
plt.title("child_mort vs class")

plt.subplot(1,2,2)
sns.boxplot(data=df, x="Class", y="income")
plt.title("income vs class")

plt.show()
```

```python
#dünya haritası üzerinde ülkeleri sınıflandırmak için plotly kütüphanesini kullanıyoruz.
import plotly.express as px
```

```python
# pca_df2 veri çerçevesine, orijinal df veri çerçevesindeki 'country' sütununu ekliyoruz. 
# Bu sayede, her bir ülkenin hangi kümeye ait olduğunu görebiliriz.
pca_df2.insert(0, column= "Country", value = df['country'])
```

```python
# KMeans modelinin her bir veri noktasını hangi kümeye atadığını gösteren etiketleri (labels) pca_df2 veri çerçevesine ekliyoruz.
pca_df2['Class'] = labels
```

```python
#burda, pca_df2 veri çerçevesindeki 'Class' sütunundaki sayısal değerleri, daha anlamlı olan string değerlerle değiştiriyoruz.
pca_df2.loc[pca_df2['Class'] == 1, 'Class'] = "Budget Needed"
pca_df2.loc[pca_df2['Class'] == 2, 'Class'] = "In Between"
pca_df2.loc[pca_df2['Class'] == 0, 'Class'] = "No Budget Needed"
```

```python
# pca_df2 veri çerçevesindeki 'Country' ve 'Class' sütunlarını kullanarak bir choropleth haritası oluşturuyoruz.
# Daha detaylı açıklama: Choropleth haritası, farklı bölgeleri renklerle temsil eden bir harita türüdür. 
# Bu haritada, her bir ülke, KMeans algoritması ile belirlenen sınıflarına göre renklendirilir. 
# "Budget Needed" sınıfına ait ülkeler kırmızı, 
# "In Between" sınıfına ait ülkeler sarı ve "No Budget Needed" sınıfına ait ülkeler yeşil renkle gösterilir. 
# Bu sayede, dünya genelinde hangi ülkelerin bütçe ihtiyacı olduğunu görselleştirebiliriz.
fig = px.choropleth(
    pca_df2[['Country', 'Class']],
    locationmode = "country names",
    locations = "Country",
    title = "Needed Budget by Country",
    color = pca_df2['Class'],
    color_discrete_map= {
                        "Budget Needed" : "Red",
                        "In Between" : "Yellow",
                        "No Budget Needed": "Green"
    })
fig.update_geos(fitbounds = "locations", visible = True)
fig.show()
```
